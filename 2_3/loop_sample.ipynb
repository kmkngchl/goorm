```python
fruits = ["사과","배","메론"]

for idx, fruit in enumerate(fruits, start=1):
    print(idx, fruit)

prices = [1000, 2000, 3000]
for f, p in zip(fruits, prices):
    print(f"{f}: {p}원")

fruits = {"사과": 1000, "배":2000, "메론": 3000}
for key, value in fruits.items():
    print(key, ":", value, "원")

fruits = [{"name": "사과", "price":1000}, {"name": "배", "price":2000}, {"name": "메론", "price":3000}]
for i in fruits:
    i["price"] += 1000

print(fruits)
```
