```python
# if 

score = 85
types = "우수교육생"

if score >= 90 and types == "우수교육생":   # and 두가지조건이 부합해야 A 
    grade = "A"
elif score >= 80 or types != "우수교육생":  # 조건이 둘 다 참일 경우. 제일 먼저 만족된 조건문에서 실행이 됨. # != 같지않다.
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(grade)
```

```python
# match 매치케이스

light = "빨강"

match light:
    case "빨강":
        print("정지")
    case "노랑":
        print("주의")
    case "초록":
        print("출발")
    case _:  # _그 외의 모든 값을 여기다가 가져다쓴다. 예외처리
        print("잘못된 신호")
```

```python
# 삼항연산자 한줄에다가 해당값, 조건문 예외값 다 들어가있어. 너무 복잡한 코드 한줄에 넣으면 안돼,간단한 조건문만 넣어야 가독성이 좋다.
age = 19
status = "성인" if age >= 20 else "미성년자"
print(status)
```
