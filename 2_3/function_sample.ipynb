```python
def calc(a, op, b=2):   # b=2 라고 하면 사용자가 입력하지 않아도 2가 기본값으로 들어감, 기본값이 디폴트 되어있으면 맨 뒤로 작성
    match op:
        case "+":
            return a+b
        case "-":
            return a-b
        case "*":
            return a*b
        case "/":
            return a/b
        case _:
            return "Error"

print(calc(10, "+"))
print(calc(10, "/", b=5))  # b=5라고 명시를 해주려면 맨뒤로 작성.
```
