```python
# 숫자형 자료형
x = 10
y = 3.14

print(x + y)
print(x - y)
print(x * y)
print(x / y)

a, b = 17, 5
print(a // b)  # 몫
print(a % b)   # 나머지
print(a ** b)  # 거듭제곱
```

```python
# 문자열 자료형
s = "hello"
print(s[0]) # 문자의 첫번쨰부터 0,1,2,3,4 ex) h = 0 e = 1 l =2 l = 3 o = 4
print(s[-1]) # o 가 출력됨
print(s[0:3]) # : 길이지정하는것 0~2 시작위치는 포함, 끝위치는 포함x 
print(s[::-1]) # :: 읽는 순서
print(s.upper(), s.lower()) # upper케이스 대문자
print(s.count("l")) # count 문자세기 "l" 의 숫자를세줌
print() # replace 바꿔줌 내가 원하는 문자로 
s= s.replace("ello", "alo")
print(s)
```

```python
# list 튜플도 거의동일하게 사용함

nums = [5, 3, 8, 1]

nums.append(10)  # 기본함수 숫자 추가 append(끝에추가됨) 한번 추가,삭제하면 영구적으로 추가되거나 삭제.
print(nums)
nums.insert(1, 99) # insert 3앞에 99가 추가됨 0으로 바꾸면 5앞에 추가됨
print(nums)
nums.remove(5) # remove 삭제
print(nums)
del nums[3] # 0,1,2,3 맞게 숫자가 삭제.
print(nums)
del nums[-1] # 제일 마지막 숫자 삭제
print(nums)
print(nums.index(99)) # index 특정 숫자 찾기 자릿수에가 출력됨
del nums[nums.index(99)] # 이런식으로도 99라는 숫자를 찾아 삭제가 가능함 
print(nums.count(3)) # 해당 숫자 갯수세기

nums = [5, 3, 8, 1, 10] 
nums.sort() # 정렬시키기 리스트에서 가장 중요한 함수. 오름차순
print(nums)
nums.sort(reverse=True) # 내림차순 리버스를 활성화 시키겠다. 
print(nums)

print(len(nums)) # 리스트 길이로 출력, 총 5개의 내용이 들어가있다.
print(sum(nums), max(nums), min(nums)) # 전체 더하기, 최대값, 최소값 찾기
```

```python
# dict

user = {"name": "한성주", "age": 30} # "user" = 키 값 , "한성주" 밸류(그의대응값)
print(user["name"])
print(user.get("email", "미등록")) # get 등록되어 있는 기본값 불러오기 없으면 미등록으로 불러오기.
user["email"] = "미등록" #새로운 키값 추가
print(user)
user["age"] = 31
print(user)
user.update({"age":32, "city":"제주도"}) # 한번에 변경
print(user)
```
